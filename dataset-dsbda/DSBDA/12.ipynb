```python
import pandas as pd

df = pd.read_csv("adult.csv", header=None)

df.columns = ['age', 'workclass', 'fnlwgt', 'education', 'education-num',
              'marital-status', 'occupation', 'relationship', 'race', 'sex',
              'capital-gain', 'capital-loss', 'hours-per-week', 'native-country', 'income']

# Remove spaces in values
df = df.apply(lambda x: x.str.strip() if x.dtype == "object" else x)

print(df.head())
```

```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Subset for USA people
usa = df[df['native-country'] == 'United-States']

# Subset for Male people
male = df[df['sex'] == 'Male']

# Subset for White race
white = df[df['race'] == 'White']

print("USA Subset")
print(usa.head())

print("Male Subset")
print(male.head())

print("White Race Subset")
print(white.head())
```

```python
# Merge USA and Male subsets
merged_data = pd.concat([usa, male])

print(merged_data.head())
print("Total Rows:", merged_data.shape[0])
```

```python
# Adult dataset has no customer ratings column.
# So sorting using education-num (numeric rating-like field)

sorted_data = df.sort_values(by='education-num')

print(sorted_data[['education', 'education-num']].head())
```

```python
# Descending order

sorted_data_desc = df.sort_values(by='education-num', ascending=False)

print(sorted_data_desc[['education', 'education-num']].head())
```

```python
# Convert rows into columns

transpose_data = df.transpose()

print(transpose_data.head())
```

```python
# Convert wide data into long format

melted_data = pd.melt(df,
                      id_vars=['sex'],
                      value_vars=['age', 'hours-per-week'],
                      var_name='Attribute',
                      value_name='Value')

print(melted_data.head())
```

```python
# Convert long format back to wide format

wide_data = melted_data.pivot_table(index='sex',
                                    columns='Attribute',
                                    values='Value',
                                    aggfunc='mean')

print(wide_data)
```
