```python
import pandas as pd
import numpy as np

# Load dataset WITHOUT header
df = pd.read_csv("hepatitis.csv", header=None)

# Assign correct column names (standard Hepatitis dataset)
df.columns = [
    'CLASS','AGE','SEX','STEROID','ANTIVIRALS','FATIGUE','MALAISE',
    'ANOREXIA','LIVER_BIG','LIVER_FIRM','SPLEEN_PALPABLE','SPIDERS',
    'ASCITES','VARICES','BILIRUBIN','ALK_PHOSPHATE','SGOT',
    'ALBUMIN','PROTIME','HISTOLOGY'
]

print(df.head())
print(df.columns)
```

```python
# Replace '?' with NaN
df.replace('?', np.nan, inplace=True)

# Convert numeric
df = df.apply(pd.to_numeric, errors='coerce')

# Drop missing values
df.dropna(inplace=True)

# Subsets
male = df[df['SEX'] == 1]
female = df[df['SEX'] == 2]
```

```python
# Merge male and female datasets
merged_data = pd.concat([male, female])

print("Merged Shape:", merged_data.shape)
```

```python
sorted_data = df.sort_values(by=['AGE', 'SGOT', 'PROTIME'])

print(sorted_data[['AGE','SGOT','PROTIME']].head())
```

```python
transpose_data = df.transpose()

print(transpose_data.head())
```

```python
melted_data = pd.melt(df,
                      id_vars=['SEX'],
                      value_vars=['AGE','SGOT','PROTIME'],
                      var_name='Attribute',
                      value_name='Value')

print(melted_data.head())
```

```python
wide_data = melted_data.pivot_table(index='SEX',
                                    columns='Attribute',
                                    values='Value',
                                    aggfunc='mean')

print(wide_data)
```
