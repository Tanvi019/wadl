```python
from google.colab import files
uploaded = files.upload()

# Perform the following operations using Python on Movie dataset
# File Used: movies_metadata.csv

import pandas as pd

# Load dataset (low_memory=False avoids mixed datatype warning)
df = pd.read_csv("movies_metadata.csv")

# Display first 5 rows
print(df.head())

# Display column names
print(df.columns)
```

```python
# Create subset for English movies
english_movies = df[df['original_language'] == 'en']

# Create subset for French movies
french_movies = df[df['original_language'] == 'fr']

# Create subset for Hindi movies
hindi_movies = df[df['original_language'] == 'hi']

# Show first 5 rows of each subset
print("English Movies")
print(english_movies.head())

print("French Movies")
print(french_movies.head())

print("Hindi Movies")
print(hindi_movies.head())
```

```python
# Merge English and French movies into one dataset
merged_movies = pd.concat([english_movies, french_movies])

# Display merged data
print(merged_movies.head())

# Total rows after merge
print("Total Rows:", merged_movies.shape[0])
```

```python
# Sort movies by vote_average (customer ratings) in ascending order
sorted_movies = df.sort_values(by='vote_average')

print(sorted_movies[['title', 'vote_average']].head())
```

```python
# Sort movies by vote_average in descending order
sorted_movies_desc = df.sort_values(by='vote_average', ascending=False)

print(sorted_movies_desc[['title', 'vote_average']].head())
```

```python
# Transpose rows into columns and columns into rows
transpose_data = df.transpose()

print(transpose_data.head())
```

```python
# Convert wide data into long format
# We use title as identifier and convert budget + revenue columns

melted_data = pd.melt(df,
                      id_vars=['title'],
                      value_vars=['budget', 'revenue'],
                      var_name='Financial_Type',
                      value_name='Amount')

print(melted_data.head())
```

```python
# Convert long format back into wide format

wide_data = melted_data.pivot_table(index='title',
                                    columns='Financial_Type',
                                    values='Amount',
                                    aggfunc='first')

print(wide_data.head())
```
