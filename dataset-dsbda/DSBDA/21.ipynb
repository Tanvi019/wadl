```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset (no header, with spaces)
df = pd.read_csv("adult.csv", header=None, skipinitialspace=True)

# Assign column names
df.columns = [
    'age','workclass','fnlwgt','education','education-num',
    'marital-status','occupation','relationship','race','sex',
    'capital-gain','capital-loss','hours-per-week','native-country','income'
]

# Basic cleaning
df.replace('?', pd.NA, inplace=True)
df.dropna(inplace=True)

# Strip spaces
#for col in df.select_dtypes(include='object').columns:
    #df[col] = df[col].str.strip()
```

```python

# Histogram
plt.figure(figsize=(8,5))

plt.hist(df['age'], bins=10)

plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python

plt.figure()
plt.plot(df['age'], df['hours-per-week'], 'o')
plt.title("Dot Plot: Age vs Working Hours")
plt.xlabel("Age")
plt.ylabel("Hours per Week")
plt.show()
# Dot Plot
```

```python



# Count income classes
income_count = df['income'].value_counts()
# Bar Plot
plt.figure(figsize=(6,5))

plt.bar(income_count.index, income_count.values)

plt.title("Income Class Distribution")
plt.xlabel("Income")
plt.ylabel("Count")

plt.show()

```

```python


# Line Chart
plt.figure(figsize=(10,5))

plt.plot(df['hours-per-week'].head(50))

plt.title("Line Chart of Hours Per Week")
plt.xlabel("Records")
plt.ylabel("Hours Per Week")

plt.show()
```

```python
# Create subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# ==========================================
# Histogram
# ==========================================

sns.histplot(
    df['age'],
    bins=20,
    color='lightblue',
    ax=ax1
)

ax1.set_title("Histogram of Age")
ax1.set_xlabel("Age")
ax1.set_ylabel("Frequency")

# ==========================================
# Box Plot + Scatter Plot
# ==========================================

# Box Plot
sns.boxplot(
    x=df['age'],
    color='lightgreen',
    ax=ax2
)

# Scatter Plot
sns.stripplot(
    x=df.sample(500)['age'],
    color='red',
    alpha=0.5,
    ax=ax2
)

ax2.set_title("Box Plot + Scatter Plot")
ax2.set_xlabel("Age")

# ==========================================
# Show Plot
# ==========================================

plt.tight_layout()
plt.show()
```
