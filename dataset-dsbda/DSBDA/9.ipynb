```python
# Perform the following operations using Python on Facebook metrics dataset

import pandas as pd

# Load Dataset
df = pd.read_csv("dataset_Facebook.csv", sep=';')

# Display first 5 rows
print(df.head())
```

```python
# Subset for Photo posts
photo_posts = df[df['Type'] == 'Photo']

# Subset for Status posts
status_posts = df[df['Type'] == 'Status']

# Subset for Video posts
video_posts = df[df['Type'] == 'Video']

print(photo_posts.head())
print(status_posts.head())
print(video_posts.head())
```

```python
# Merge Photo and Status posts
merged_data = pd.concat([photo_posts, status_posts])

print(merged_data.head())
print(status_posts.shape[0])
print("Total rows after merge:", merged_data.shape[0])
```

```python
# Ascending order
sorted_data = df.sort_values(by='Page total likes')

print(sorted_data[['Page total likes', 'Type']].head())
```

```python
# Descending order
sorted_data_desc = df.sort_values(by='Page total likes', ascending=False)

print(sorted_data_desc[['Page total likes', 'Type']].head())
```

```python
# Transpose dataset
transpose_data = df.transpose()

print(transpose_data.head())
```

```python
# Convert selected columns into long format
melted_data = pd.melt(df,
                      id_vars=['Type'],
                      value_vars=['like', 'share', 'comment'],
                      var_name='Metrics',
                      value_name='Count')

print(melted_data.head())
```

```python
# Convert long format back to wide format
wide_data = melted_data.pivot_table(index='Type',
                                    columns='Metrics',
                                    values='Count',
                                    aggfunc='mean')

print(wide_data)
```
