```python
from google.colab import files
uploaded = files.upload()


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("adult.csv", header=None, skipinitialspace=True)

# Assign column names
df.columns = [
    'age','workclass','fnlwgt','education','education-num',
    'marital-status','occupation','relationship','race','sex',
    'capital-gain','capital-loss','hours-per-week','native-country','income'
]

# Basic cleaning
df.replace('?', pd.NA, inplace=True)
df.dropna(inplace=True)

# Remove extra spaces
#for col in df.select_dtypes(include='object').columns:
    #df[col] = df[col].str.strip()
```

```python


# Histogram
plt.figure()

plt.hist(df['age'], bins=10)

plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python

counts = df['income'].value_counts()

plt.figure()
plt.pie(counts, labels=counts.index, autopct='%1.1f%%')
plt.title("Income Distribution")
plt.show()
```

```python
plt.figure()
sns.boxplot(x=df['hours-per-week'])
plt.title("Box Plot of Working Hours")
plt.show()

```

```python
plt.figure()
plt.scatter(df['age'], df['hours-per-week'])
plt.title("Scatter: Age vs Working Hours")
plt.xlabel("Age")
plt.ylabel("Hours per Week")
plt.show()
```

```python
#add box to scatter polt
# Create Joint Plot
sns.jointplot(x='age',y='hours-per-week',data=df,kind='scatter',marginal_kws=dict(bins=15, fill=True))

plt.show()
```
