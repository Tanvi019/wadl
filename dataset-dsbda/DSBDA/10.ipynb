```python
from google.colab import files
uploaded = files.upload()


import pandas as pd

# Load file without header
df = pd.read_csv("Iris.csv", header=None)

# Assign proper column names
df.columns = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm', 'Species']

print(df.head())
```

```python
setosa = df[df['Species'] == 'Iris-setosa']
versicolor = df[df['Species'] == 'Iris-versicolor']
virginica = df[df['Species'] == 'Iris-virginica']

print(setosa.head())
print(versicolor.head())
print(virginica.head())
```

```python
merged_data = pd.concat([setosa, versicolor])

print(merged_data.head())
```

```python
sorted_data = df.sort_values(by='PetalLengthCm')

print(sorted_data.head())
```

```python
print(df.transpose())
```

```python
melted_data = pd.melt(df,
                      id_vars=['Species'],
                      value_vars=['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm'],
                      var_name='Features',
                      value_name='Value')

print(melted_data.head())
```

```python
#(l) Casting to wide format (Pivot)
pivot_df = melted_data.pivot_table(index='Species',
                                columns='Features',
                                values='Value',
                                aggfunc='mean')

pivot_df
```
